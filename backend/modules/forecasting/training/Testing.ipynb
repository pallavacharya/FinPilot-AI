```python
import matplotlib.pyplot as plt
import pandas as pd
from src.forecast import get_chart_data

data = get_chart_data('Accessories')
h = pd.DataFrame(data['history']); h['date'] = pd.to_datetime(h['date'])
f = pd.DataFrame(data['forecast']); f['date'] = pd.to_datetime(f['date'])

plt.figure(figsize=(12, 4))
plt.plot(h['date'], h['sales'], label='Actual sales')
plt.plot(f['date'], f['predicted'], label='Forecast', color='orange')
plt.fill_between(f['date'], f['lower'], f['upper'], color='orange', alpha=0.2, label='Likely range')
plt.title('Chairs: weekly sales and forecast')
plt.legend()
plt.show()
```
